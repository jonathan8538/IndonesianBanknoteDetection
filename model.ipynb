```python
model = YOLO('yolov8n.pt')
results = model.train(
    data='/content/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    save_period=10,
    val=True,
    workers=2,
    device='0'
)
```

```python
model = YOLO('/content/runs/detect/train3/weights/best.pt')
results = model.val(data='/content/data2.yaml', split='test')

class_names = model.names
precisions = results.box.p
recalls = results.box.r
map50s = results.box.ap50
map5095s = results.box.ap

print(f"{'Class':<10} {'Precision':<10} {'Recall':<10} {'mAP@0.5':<10} {'mAP@0.5:0.95'}")
print("="*60)
for i in range(len(class_names)):
    print(f"{class_names[i]:<10} {precisions[i]:<10.4f} {recalls[i]:<10.4f} {map50s[i]:<10.4f} {map5095s[i]:<10.4f}")
```
